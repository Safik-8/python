```python
import pandas as pd

data = {
    "name": ["A", "B", "C", "D", "E"],
    "age": [20, 25, 30, 35, 40],
    "salary": [30000, 40000, 50000, 100000000000, 10],
    "department": ["IT", "HR", "it", "HR","IT"],
    "gender":["M","M","F","M","M"],
    "experience":["s", "m","b","xs","lg"]
}

df = pd.DataFrame(data)
print(df)
```

```python
import matplotlib.pyplot as plt

plt.plot(df["age"], df["salary"])
plt.show()
```

```python
import seaborn as sns

sns.lineplot(data=df, x="age", y="salary")
```

```python
sns.scatterplot(
    data=df,
    x="age",
    y="salary"
)
```

```python
sns.scatterplot(
    data=df,
    x="age",
    y="salary",
    hue="department"
)
```

```python
sns.scatterplot(
    data=df,
    x="age",
    y="salary",
    hue="department",
    style="gender"
)
```

```python
sns.scatterplot(
    data=df,
    x="age",
    y="salary",
    size="experience"
)
```

```python
sns.histplot(data=df, x="salary")
```

```python
sns.histplot(
    data=df,
    x="salary",
    kde=True
)
```

```python
sns.barplot(
    data=df,
    x="department",
    y="salary"
)
```

```python
sns.boxplot(
    data=df,
    x="salary"
)
```

```python
sns.boxplot(
    data=df,
    x="department",
    y="salary"
)
```

```python
sns.countplot(data=df, x="department")
```
