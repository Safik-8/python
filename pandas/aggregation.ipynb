```python
import pandas as pd

data = {
    "city": ["Ahmedabad", "Ahmedabad", "Ahmedabad",
             "Mumbai", "Mumbai", "Mumbai",
             "Delhi", "Delhi", "Delhi"],

    "product": ["Laptop", "Mobile", "Laptop",
                "Laptop", "Mobile", "Mobile",
                "Laptop", "Mobile", "Laptop"],

    "sales": [50000, 20000, 55000,
              60000, 25000, 30000,
              52000, 22000, 58000]
}

df = pd.DataFrame(data)

print(df)


pd.pivot_table(
    df,
    index="city",
    columns="product",
    values="sales",
    aggfunc="mean"
)
```

```python

data = {
    "name": ["A", "B", "C", "D", "E", "F", "G", "H"],
    "city": [
        "Ahmedabad", "Ahmedabad", "Ahmedabad",
        "Mumbai", "Mumbai",
        "Delhi", "Delhi", "Delhi"
    ],
    "score": [80, 90, 70, 60, 80, 90, 70, 80]
}

df = pd.DataFrame(data)

print(df)
```

```python
df.groupby("city")
```

```python
df.groupby("city")["score"].median()
```

```python
df.groupby("city")["city"].median()
```

```python
df.groupby("city")["score"].agg(["mean", "max", "count", "median"])
```

```python
df.groupby("city").agg({
"score": "mean",
# "age": "max",
"name": "count",
})
```

```python
df.groupby("city")["score"].mean().sort_values(ascending=False)
```

```python
df.pivot_table(
values="sales",
index="region", # rows
columns="product", # columns
aggfunc="sum", # how to combine
fill_value=0, # replace missing combos with 0
)
```
